```python
ls = ['audio_song', 'audio_speech']
test_disgust = os.path.join(data_path, 'disgust_audio_test')
train_disgust = os.path.join(data_path, 'disgust_audio_train')
test_surprised = os.path.join(data_path, 'surprised_audio_test')
train_surprised = os.path.join(data_path, 'surprised_audio_train')

for dir in ls:
    pth = os.path.join(data_path, dir)
    for actor in os.listdir(pth):
        pth2 = os.path.join(pth, actor)
        if (int(actor[6:])%4 == 0):
            for file in os.listdir(pth2):
                if (file[7] == '7'):
                    shutil.move(os.path.join(pth2, file), os.path.join(test_disgust, file))
                if (file[7] == '8'):
                    shutil.move(os.path.join(pth2, file), os.path.join(test_surprised, file))
        else:
            for file in os.listdir(pth2):
                if (file[7] == '7'):
                    shutil.move(os.path.join(pth2, file), os.path.join(train_disgust, file))
                if (file[7] == '8'):
                    shutil.move(os.path.join(pth2, file), os.path.join(train_surprised, file))
```
